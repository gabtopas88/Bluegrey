# Indicator column count check

Brief check of how many columns exist before and after `compute_technical_indicators()`.

```python
from pathlib import Path
import importlib.util
import sys

# Find the repo root from wherever Jupyter was started
repo_root = Path.cwd().resolve()
while not (repo_root / "research" / "Mateo" / "ML_predictor" / "ML_predictor-CODEX_SUGS.py").exists():
    repo_root = repo_root.parent

# Let ML_predictor-CODEX_SUGS.py import repo modules such as strategies.base
sys.path.insert(0, str(repo_root))

# Import ML_predictor-CODEX_SUGS.py despite the hyphen in the filename
script_path = repo_root / "research" / "Mateo" / "ML_predictor" / "ML_predictor-CODEX_SUGS.py"
spec = importlib.util.spec_from_file_location("ml_predictor_codex_sugs", script_path)
ml_predictor = importlib.util.module_from_spec(spec)
spec.loader.exec_module(ml_predictor)

MLPredictorStrategy = ml_predictor.MLPredictorStrategy
```

```python
from ML_predictor_CODEX_SUGS import MLPredictorStrategy
```

```python
strategy = MLPredictorStrategy(params={
    "symbol": "ECH",
    "start_date": "2026-01-01",
    "end_date": "2026-02-01",
    "training_years": 0,
    "indicator_warmup_years": 0,
    "verbose": True,
})

# Reproduce prepare_dataset() only up to the indicator step
df = strategy.load_stored_dataset()
print("\nAfter load_stored_dataset")
display(df)
df = strategy.adjust_dataset(df)
print("After adjust_dataset")
display(df)
df = strategy.assign_signal(df)
print("After assign_signal")
display(df)

# Count columns immediately before and after indicators are added
before_cols = df.columns.tolist()
print(f"Columns before indicators: {len(before_cols)}")
print(f"\nColumns before indicators:\n{before_cols}")
df_with_indicators = strategy.compute_technical_indicators(df.copy())
print("After compute_technical_indicators")
display(df)
after_cols = df_with_indicators.columns.tolist()
print(f"Columns after indicators:  {len(after_cols)}")
print(f"Columns added:             {len(after_cols) - len(before_cols)}")
```
