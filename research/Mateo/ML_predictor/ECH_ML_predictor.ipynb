## 4 - 3D grid search over promising ECH parameters

This runs a stack of 2D `n_features` x `signal_threshold` grid searches, one slice per `training_years` value, over `2022-01-01` to `2026-01-01`. The scope is intentionally fine around the already-promising regions, while excluding consistently weak areas such as `training_years < 5` and very high signal thresholds.

It will be split into a first part with `training_years = [5 to 10]` and a second part with `training_years = [11 to 17]`

### 4a - First 3D grid search slices for ECH

This starts the effective 3D grid search over `2022-01-01` to `2026-01-01`, covering the first training-window batch. These slices cover `training_years = 5..10`.

```python
import pandas as pd

# Fine 3D search scope. Total cases = len(n_features) * len(signal_thresholds) * len(training_years).
ech_3d_n_features = list(range(3, 17))
ech_3d_signal_thresholds = [
    0.0000,
    0.0005,
    0.0010,
    0.0015,
    0.0020,
    0.0025,
    0.0030,
    0.0035,
    0.0040,
    0.0045,
    0.0050,
]
ech_3d_training_years = list(range(5, 11))

ech_3d_total_cases = (
    len(ech_3d_n_features)
    * len(ech_3d_signal_thresholds)
    * len(ech_3d_training_years)
)

print(f"ECH 3D grid search will run {ech_3d_total_cases:,} cases")
print(f"n_features: {ech_3d_n_features}")
print(f"signal_thresholds: {ech_3d_signal_thresholds}")
print(f"training_years: {ech_3d_training_years}")
```

```python
# Each loop iteration is a 2D heatmap slice at a fixed training_years value.
# The result dictionaries are intentionally preserved if this cell is interrupted and rerun in the same kernel.
try:
    ech_3d_grid_results
except NameError:
    ech_3d_grid_results = {}

try:
    ech_3d_heatmaps
except NameError:
    ech_3d_heatmaps = {}

for training_years in ech_3d_training_years:
    if training_years in ech_3d_grid_results:
        print(f"Skipping training_years={training_years}: already completed")
        continue

    print(f"Running ECH 3D grid slice: training_years={training_years}")
    grid_df, heatmap = parameter_grid_search(
        symbol="ECH",
        asset_class="STK",
        var_1="n_features",
        list_1=ech_3d_n_features,
        var_2="signal_threshold",
        list_2=ech_3d_signal_thresholds,
        training_years=training_years,
        position_size=1.0,
        start_date="2022-01-01",
        end_date="2026-01-01",
        n_jobs=14,
        verbose=10,
    )

    ech_3d_grid_results[training_years] = grid_df
    ech_3d_heatmaps[training_years] = heatmap

play_sound()
```

### 4b - Remaining 3D grid search slices for ECH

This continues the effective 3D grid search over `2022-01-01` to `2026-01-01`, using separate variable names so it does not overwrite the currently running `training_years = 5..10` run. These slices cover `training_years = 11..16`.

```python
import pandas as pd

# Fine 3D search scope. Total cases = len(n_features) * len(signal_thresholds) * len(training_years).
ech_3d_n_features = list(range(3, 17))
ech_3d_signal_thresholds = [
    0.0000,
    0.0005,
    0.0010,
    0.0015,
    0.0020,
    0.0025,
    0.0030,
    0.0035,
    0.0040,
    0.0045,
    0.0050,
]
ech_3d_training_years = list(range(11, 17))

ech_3d_current_run_cases = (
    len(ech_3d_n_features)
    * len(ech_3d_signal_thresholds)
    * len(ech_3d_training_years)
)

ech_3d_total_cases = (
    len(ech_3d_n_features)
    * len(ech_3d_signal_thresholds)
    * len(range(5, 17))
)

print(f"ECH 3D grid search will run {ech_3d_current_run_cases:,} new cases")
print(f"Full ECH 3D grid search target is {ech_3d_total_cases:,} cases")
print(f"n_features: {ech_3d_n_features}")
print(f"signal_thresholds: {ech_3d_signal_thresholds}")
print(f"training_years: {ech_3d_training_years}")
```

```python
# Each loop iteration is a 2D heatmap slice at a fixed training_years value.
# The result dictionaries are intentionally preserved if this cell is interrupted and rerun in the same kernel.
try:
    ech_3d_grid_results
except NameError:
    ech_3d_grid_results = {}

try:
    ech_3d_heatmaps
except NameError:
    ech_3d_heatmaps = {}

for training_years in ech_3d_training_years:
    if training_years in ech_3d_grid_results:
        print(f"Skipping training_years={training_years}: already completed")
        continue

    print(f"Running ECH 3D grid slice: training_years={training_years}")
    grid_df, heatmap = parameter_grid_search(
        symbol="ECH",
        asset_class="STK",
        var_1="n_features",
        list_1=ech_3d_n_features,
        var_2="signal_threshold",
        list_2=ech_3d_signal_thresholds,
        training_years=training_years,
        position_size=1.0,
        start_date="2022-01-01",
        end_date="2026-01-01",
        n_jobs=14,
        verbose=10,
    )

    ech_3d_grid_results[training_years] = grid_df
    ech_3d_heatmaps[training_years] = heatmap

play_sound()
```

### 4c - Combine completed 3D grid search slices for ECH

This combines the completed `training_years = 5..10` and `training_years = 11..16` 2D grid-search slices into one effective 3D grid-search result table.

```python
# Combine the completed 2D slices into one effective 3D grid-search table.
ech_3d_completed_dfs = [
    grid_df
    for _, grid_df in sorted(ech_3d_grid_results.items())
    if grid_df is not None and not grid_df.empty
]

if not ech_3d_completed_dfs:
    raise RuntimeError("No completed ECH 3D grid-search slices found yet.")

ech_3d_grid_df = (
    pd.concat(ech_3d_completed_dfs, ignore_index=True)
    .sort_values(["training_years", "n_features", "signal_threshold"])
    .reset_index(drop=True)
)

ech_3d_best_overall = (
    ech_3d_grid_df
    .sort_values("net_return_pct", ascending=False)
    .head(30)
    .reset_index(drop=True)
)

ech_3d_best_by_training_years = (
    ech_3d_grid_df
    .sort_values("net_return_pct", ascending=False)
    .groupby("training_years", as_index=False)
    .head(3)
    .sort_values(["training_years", "net_return_pct"], ascending=[True, False])
    .reset_index(drop=True)
)

print(f"Completed {len(ech_3d_grid_df):,} / {ech_3d_total_cases:,} planned cases")
print("Top 30 parameter combinations overall:")
display(ech_3d_best_overall)

print("Top 3 parameter combinations per training_years slice:")
display(ech_3d_best_by_training_years)
```

```python
ech_3d_best_overall_top10 = ech_3d_best_overall.head(10)[["symbol", "start_date", "end_date", "signal_threshold", "n_features", "training_years"]].to_dict(orient="records")

for idx, params in enumerate(ech_3d_best_overall_top10, start=1):
    print(f"Running top {idx}/10: n_features={params['n_features']}, signal_threshold={params['signal_threshold']}, training_years={params['training_years']}")
    display(params)
    run_single_case(**params)
```
